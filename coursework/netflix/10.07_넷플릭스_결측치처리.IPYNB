```python
import pandas as pd
import os
import urllib.request

print("pandas 버전:", pd.__version__)
print("현재 작업 폴더:", os.getcwd())
```

```python
# 저장할 폴더 지정 (notebooks 기준 한 단계 위로 올라간 뒤 raw\netflex_project)
data_dir = os.path.join("..", "raw", "netflex_project")
os.makedirs(data_dir, exist_ok=True)   # 폴더가 없으면 자동 생성

file_name = os.path.join(data_dir, "netflix_titles.csv")
url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/master/data/2021/2021-04-20/netflix_titles.csv"

if not os.path.exists(file_name):
    print("다운로드를 시작합니다...")
    urllib.request.urlretrieve(url, file_name)
    print("다운로드 완료")
else:
    print("파일이 이미 있습니다")

print("저장 위치:", os.path.abspath(file_name))
```

```python
df = pd.read_csv(file_name)

print(f"행 개수: {df.shape[0]}, 열 개수: {df.shape[1]}")
df.head()
```

```python
print("### 결측치 개수 ###")
print(df.isnull().sum())

print("\n### 결측치 비율(%) ###")
print((df.isnull().sum() / len(df) * 100).round(2))
```
